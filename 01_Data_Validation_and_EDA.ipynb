```python
# Step 1: Identify Variable Groups
# Helper function to select columns by prefix

def cols_with_prefix(df, prefixes):
    cols = []
    for prefix in prefixes:
        cols.extend([c for c in df.columns if c.startswith(prefix)])
    return sorted(list(set(cols)))
```

```python
# clinical Dataset

clinical_prefixes = [
    "BMX",
    "BPX",
    "BPQ",
    "DIQ",
    "LBX",
    "LBD"
]

clinical_cols = cols_with_prefix(df, clinical_prefixes)

clinical_cols += [
    "RIAGENDR",
    "RIDAGEYR",
    "RIDRETH3",
    "DMDEDUC2",
    "DMDMARTZ",
    "INDFMPIR",
    "Composite_CVD"
]

clinical_cols = [c for c in clinical_cols if c in df.columns]

clinical_df = df[clinical_cols].copy()

print(clinical_df.shape)
```

```python
# Lifestyle & Environment Dataset

life_prefixes = [
    "SMQ",
    "ALQ",
    "PAQ",
    "PAD",
    "SLQ",
    "DR1",
    "DR2",
    "DS",
    "URX",
    "OCQ",
    "HIQ"
]

life_cols = cols_with_prefix(df, life_prefixes)

life_cols.append("Composite_CVD")

life_cols = [c for c in life_cols if c in df.columns]

lifestyle_df = df[life_cols].copy()

print(lifestyle_df.shape)
```

```python
# Integrated Dataset

remove_cols = [
    "SEQN",
    "Composite_CVD"
]

integrated_df = df.drop(
    columns=[c for c in remove_cols if c in df.columns]
).copy()

integrated_df["Composite_CVD"] = df["Composite_CVD"]

print(integrated_df.shape)
```
